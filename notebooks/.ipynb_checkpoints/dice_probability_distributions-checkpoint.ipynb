# Satya CARPG: Dice Mechanics & Probability Distribution Topologies

All probability distributions in this notebook are computed **exactly** (zero Monte Carlo sampling noise) via combinatorial and dynamic-programming engines in [`src/satya_ml/dice_probabilities.py`](../src/satya_ml/dice_probabilities.py) and rendered interactively with **Plotly** (which can also serialize figures via `fig.to_json()` for an in-game Satya odds widget).

## 1. Additive & Negative Dice Pools: Why the Central Limit Theorem Forces a Bell Curve

- **$1\text{d}6$**: Flat **Uniform** distribution ($\text{kurtosis} = -1.27$).
- **$2\text{d}6$**: Symmetric **Triangular** distribution ($\text{kurtosis} = -0.63$).
- **$4\text{d}6$ and $3\text{d}6 - 2\text{d}6$ (Negative Dice)**: Rapidly converge to a **Gaussian (Normal)** distribution ($\text{skewness} = 0$, $\text{excess kurtosis} \to 0$). Notice on the right subplot how normalizing by mean and standard deviation ($Z$-score) collapses $4\text{d}6$ and $3\text{d}6 - 2\text{d}6$ onto almost the exact same bell curve.

```python
fig_clt = plot_clt_sum_pools()
fig_clt.show()
```

## 2. Success-Counting Pools ($N\text{d}K \ge T$) & Satya's Roll-Under d20 Pool

- **Small Pool + Extreme Target Number ($p = 0.2$ or $p = 0.8$)**: Strongly skewed (Poisson-like).
- **Moderate/Large Pool ($12\text{d}10 \ge 6$)**: The De Moivre–Laplace theorem pulls the Binomial distribution back into a symmetric Gaussian bell curve.
- **Right Subplot (Satya d20 Pool)**: Shows Satya's roll-under $\text{d}20 \le \text{TN}$ mechanic where a natural `1` counts as **2 successes**, creating a slight positive right-tail boost for small pools.

```python
fig_binom = plot_success_counting_and_satya_d20()
fig_binom.show()
```

## 3. Exploding Dice: Geometric Staircases & Gumbel Extreme Value Tails

- **Left Subplot (Single Exploding d6)**: Introduces strong **positive skewness**, an exponentially decaying right tail, and zero-probability gaps at multiples of $6$ ($6, 12, 18\dots$). Lowering the explosion threshold ($5\text{–}6$ vs. $6+$) directly controls the tail decay rate.
- **Right Subplot (Keep-Highest of $N$ Exploding Dice)**: Because we take $\max(d_1, \dots, d_N)$ rather than the sum, the Central Limit Theorem never applies. Instead, the pool converges to a permanently asymmetric **Gumbel (Extreme Value Type I)** distribution.

```python
fig_explode = plot_exploding_dice()
fig_explode.show()
```

## 4. Non-Additive Selection: Discrete Beta Order Statistics & Bimodal U-Curves

- **Left Subplot (Roll $5\text{d}20$, Keep $k$-th Die)**: Picking the $k$-th ordered die maps directly to a discretized $\text{Beta}(k, N-k+1)$ distribution bounded on $1..20$. Changing $k$ smoothly sweeps skewness from strongly right-skewed ($k=1$) to symmetric ($k=3$) to strongly left-skewed ($k=5$).
- **Right Subplot (Center-Distance & Median-Gated Selection)**:
  - **Focused (`closest` to $10.5$)**: Sharp tent peak in the middle.
  - **Chaotic (`furthest` from $10.5$)**: Inverts the bell curve into a **bimodal U-shaped distribution** where $1$ and $20$ are the dominant peaks and $10\text{–}11$ rarely occur.
  - **Median-Gated $3\text{d}20$**: Uses the middle die as a gate between $\max$ and $\min$, creating **two distinct humps** with a valley in the middle.

```python
fig_order = plot_order_statistics_and_chaos()
fig_order.show()
```

## 5. Two-List Cascading / Contagious Dice: Characteristic $C$ as a Watershed Repeller

Two lists of $N$ dice are rolled ($A_1\dots A_N$ and $B_1\dots B_N$). At each index $n$, the paired sum $Y_n = A_n + B_n + \text{mod}_n$ is compared to the controlling characteristic $C$:
- If $Y_n < C$, we **subtract** $\delta_{\text{under}}$ from pair $n+1$ (pushing $n+1$ *further under* $C$).
- If $Y_n > C$, we **add** $\delta_{\text{over}}$ to pair $n+1$ (pushing $n+1$ *further over* $C$).

### Why this morphs the topology:
- **Left Subplot (Feedback Strength $\delta$ at $C=7$)**: As $\delta$ increases from $0 \to 2 \to 3 \to 4$, the distribution of pairs beating $C$ morphs from a **Gaussian/Binomial Bell Curve** ($\delta=0$) $\to$ **Flat Uniform Plateau** ($\delta=2$) $\to$ **Bimodal U-Shaped Distribution** ($\delta \ge 3$).
- **Right Subplot (Controlling Characteristic $C$ at $\delta=3$)**: Moving $C$ shifts the unstable watershed divide across the dice range, morphing the curve from a **Right-Skewed J-Curve** ($C=9$) $\to$ **Symmetric Bimodal U-Curve** ($C=7$) $\to$ **Left-Skewed Reverse-J-Curve** ($C=5$).

```python
fig_watershed = plot_paired_list_watershed(num_pairs=5, sides=6)
fig_watershed.show()
```

## 6. Two-List Cascade Final Value ($Y_N$) & The Linear Margin Trap

- **Left Subplot (Non-Linear Step Carry on Final Pair $Y_N$)**:
  - With a fixed Markov step ($\pm 3$), the final pair $Y_N$ splits into **two distinct peaks** straddling $C=7$.
  - With **Asymmetric Mastery ($+4$ when over, $-1$ when under)**, positive momentum is sticky while negative momentum is easy to recover from, creating a strong high-side peak.
- **Right Subplot (The Linear Margin Trap)**:
  - If you pass the exact linear margin $(Y_n - C)$ to step $n+1$ and **count how many steps beat $C$** (gold bars), you get **Lévy's Arcsine U-Shaped distribution**.
  - However, if you pass $(Y_n - C)$ linearly and **only look at the final value $Y_N$** (red curve), the sum telescopes into $\sum (A_i + B_i) - (N-1)C$, allowing the Central Limit Theorem to collapse $Y_N$ right back into a Normal bell curve!

```python
fig_trap = plot_cascade_final_pair_and_trap(num_pairs=5, sides=6)
fig_trap.show()
```

## 7. Multiplicative Dice Pools ($d_1 \times d_2 \times d_3$): Log-Normal Heavy Tails

While adding dice invokes the Additive Central Limit Theorem ($\to$ Normal), multiplying dice invokes the **Multiplicative Central Limit Theorem ($\to$ Log-Normal)**, compressing low rolls and stretching a massive heavy tail to the right.

```python
fig_mult = plot_multiplicative_lognormal()
fig_mult.show()
```

## 8. Interactive Playground: Tune Your Own Cascading Pool

Modify the parameters below (`NUM_PAIRS`, `DIE_SIDES`, `CHARACTERISTIC_C`, `DELTA_OVER`, `DELTA_UNDER`, `MODE`) to experiment with different configurations and inspect exact moment statistics.

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# --- Tweak Parameters Here ---
NUM_PAIRS = 6          # Length of the two dice lists A and B
DIE_SIDES = 10         # Die size in each list (e.g., d6, d10, d12, d20)
CHARACTERISTIC_C = 11  # Controlling characteristic (average of 2d10 is 11)
DELTA_OVER = 4         # Added to n+1 when pair n > C
DELTA_UNDER = 4        # Subtracted from n+1 when pair n < C
MODE = "markov_step"   # Options: "markov_step", "cumulative_snowball", "linear_margin"

succ_pmf, final_y_pmf = paired_list_cascade_pmfs(
    num_pairs=NUM_PAIRS,
    sides=DIE_SIDES,
    characteristic_c=CHARACTERISTIC_C,
    delta_over=DELTA_OVER,
    delta_under=DELTA_UNDER,
    mode=MODE,
)

s_stats = compute_stats(succ_pmf)
y_stats = compute_stats(final_y_pmf)
print(
    f"Success Count Stats -> Mean: {s_stats.mean:.2f}, Std: {s_stats.std:.2f}, "
    f"Skewness: {s_stats.skewness:+.2f}, Excess Kurtosis: {s_stats.excess_kurtosis:+.2f}"
)
print(
    f"Final Pair Y_N Stats -> Mean: {y_stats.mean:.2f}, Std: {y_stats.std:.2f}, "
    f"Skewness: {y_stats.skewness:+.2f}, Excess Kurtosis: {y_stats.excess_kurtosis:+.2f}"
)

fig_custom = make_subplots(
    rows=1,
    cols=2,
    subplot_titles=(
        f"Pairs Beating C={CHARACTERISTIC_C} (out of {NUM_PAIRS}x 2d{DIE_SIDES})",
        f"Final Pair Effective Value Y_{NUM_PAIRS} ({MODE})",
    ),
)
fig_custom.add_trace(
    go.Bar(
        x=list(succ_pmf.keys()),
        y=[v * 100 for v in succ_pmf.values()],
        name="Success Count %",
        marker_color="#38bdf8",
    ),
    row=1,
    col=1,
)
fig_custom.add_trace(
    go.Bar(
        x=list(final_y_pmf.keys()),
        y=[v * 100 for v in final_y_pmf.values()],
        name="Final Pair Y_N %",
        marker_color="#f472b6",
    ),
    row=1,
    col=2,
)
fig_custom.update_layout(template="plotly_dark", paper_bgcolor="#0f172a", plot_bgcolor="#1e293b", height=420)
fig_custom.show()
```
